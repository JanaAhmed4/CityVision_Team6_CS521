```python
import pandas as pd

# Load your files
price_df = pd.read_excel("../../Cleaned Data/Cleaned_Districts.xlsx")
combined_df = pd.read_csv("updated_combined_df_with_residential_prices.csv")

# Normalize Arabic district names for better matching
def normalize_district(name):
    if pd.isna(name):
        return name
    name = name.strip().lower()
    if not name.startswith("حي "):
        name = "حي " + name
    return name

price_df["District_ar"] = price_df["District_ar"].apply(normalize_district)
combined_df["District_normalized"] = combined_df["District"].apply(normalize_district)

# Merge without overwriting existing 'Residential Average Price'
combined_df = combined_df.merge(
    price_df[["District_ar", "Residential Average Price"]],
    left_on="District_normalized",
    right_on="District_ar",
    how="left",
    suffixes=('', '_new')
)

# Only fill missing prices
combined_df["Residential Average Price"] = combined_df["Residential Average Price"].fillna(combined_df["Residential Average Price_new"])

# Drop helper columns
combined_df.drop(columns=["Residential Average Price_new", "District_ar", "District_normalized"], inplace=True)

# Save the updated file
combined_df.to_csv("3updated_combined_df_with_matched_prices.csv", index=False)
print("Prices merged without overwriting existing values.")
```

```python
# District name mappings from English names in the combined_df to Arabic names in the price_df
district_name_map = {
    "Taybah": "حي طيبة",
    "Hajar": "حي هجر",
    "Qurtoba": "حي قرطبة",
    "Industrial Area No 1": "حي الصناعية الاولى",
    "أحد": "حي احد",
    "Village2": "حي الشعلة",
    "حي المنتزة": "حي المنتزه",
    "Sinayiat Alfawaziya": "حي صناعية الفوازية",
    "حي الشفاء": "حي عبدالله فؤاد",
    "Aliskan": "حي الاسكان"
}
# Ensure to handle NaN values
combined_df["District_mapped"] = combined_df["District"].apply(
    lambda x: district_name_map.get(str(x).strip(), str(x).strip()) if pd.notna(x) else x
)

# Then continue as before, using "District_mapped" for the merge
combined_df = combined_df.merge(
    price_df[["District_ar", "Residential Average Price"]],
    left_on="District_mapped",
    right_on="District_ar",
    how="left",
    suffixes=("", "_from_price")
)

# Fill missing Residential Average Price only if it wasn't already filled
combined_df["Residential Average Price"] = combined_df["Residential Average Price"].fillna(
    combined_df["Residential Average Price_from_price"]
)

# Optional: Drop helper columns
combined_df.drop(columns=["District_mapped", "District_ar", "Residential Average Price_from_price"], inplace=True)
# Save the updated file
combined_df.to_csv("3updated_combined_df_with_matched_prices.csv", index=False)
print("Prices merged without overwriting existing values.")
```

```python
import pandas as pd
from fuzzywuzzy import fuzz
from geopy.distance import geodesic
from itertools import combinations



# Function to calculate Haversine Distance (geo distance between two locations)
def is_nearby(lat1, lon1, lat2, lon2, threshold=0.5):
    """Returns True if two locations are within 'threshold' km."""
    return geodesic((lat1, lon1), (lat2, lon2)).km <= threshold

# Function for fuzzy name matching
def is_similar(name1, name2, threshold=85):
    """Returns True if two names are at least 'threshold'% similar."""
    return fuzz.token_sort_ratio(name1, name2) >= threshold

# Identify possible duplicate restaurants
duplicates = []

for (i, row1), (j, row2) in combinations(df.iterrows(), 2):
    if is_similar(row1["Name"], row2["Name"]) and is_nearby(row1["Latitude"], row1["Longitude"], row2["Latitude"], row2["Longitude"]):
        duplicates.append({
            "Duplicate_1_Index": i,
            "Duplicate_2_Index": j,
            "Duplicate_1_ID": row1["ID"],
            "Duplicate_2_ID": row2["ID"],
            "Name_1": row1["Name"],
            "Name_2": row2["Name"],
            "Latitude_1": row1["Latitude"],
            "Longitude_1": row1["Longitude"],
            "Latitude_2": row2["Latitude"],
            "Longitude_2": row2["Longitude"]
        })

# Create a DataFrame of duplicate pairs
duplicates_df = pd.DataFrame(duplicates)

# Print detected duplicates
print(f"Found {len(duplicates_df)} potential duplicate restaurants.")
print(duplicates_df.head())
```

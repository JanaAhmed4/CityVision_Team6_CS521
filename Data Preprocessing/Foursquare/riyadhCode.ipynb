```python
import pandas as pd

# Load the CSV file
file_path = "../../Cleaned Data/fullcoffeeshops.csv"  
business_data = pd.read_csv(file_path)

restaurants_df = pd.read_csv("riyadh_restaurants.csv")
coffeeshops_df = pd.read_csv("riyadh_coffeeshops.csv")

combined_df = pd.concat([coffeeshops_df, restaurants_df], ignore_index=True)

# Remove records where Popularity is 'No popularity data'
combined_df = combined_df[combined_df['Popularity'] != 'No popularity data']
```

#### Duplicates

```python
combined_df = combined_df.drop_duplicates(subset=['Business ID'])
```

```python
combined_df
```

#### Variance

```python
# Convert 'Popularity' column to numeric (ignore errors for non-numeric values)
combined_df["Popularity"] = pd.to_numeric(combined_df["Popularity"], errors="coerce")

# Calculate variance
variance = combined_df["Popularity"].var()

# Print result
print(f"Variance of Popularity: {variance}")
```

#### After scaling

```python
from sklearn.preprocessing import MinMaxScaler


scaler = MinMaxScaler()
popularity_scaled = scaler.fit_transform(combined_df[["Popularity"]])
combined_df["Popularity_Scaled"] = popularity_scaled
scaled_variance = combined_df["Popularity_Scaled"].var()

# Print the variance of the scaled values
print(f"Variance of Scaled Popularity: {scaled_variance}")
```
